```python
#import library
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.datasets import fashion_mnist
import matplotlib.pyplot as plt
```

```python
# Memuat data Fashion-MNIST
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

# Label nama kelas pada Fashion-MNIST
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

# Mempersiapkan data input untuk CNN
# CNN membutuhkan dimensi 4D: (jumlah_gambar, tinggi, lebar, channel)
x_train = x_train.reshape(x_train.shape[0], 28, 28, 1).astype('float32') / 255.0
x_test = x_test.reshape(x_test.shape[0], 28, 28, 1).astype('float32') / 255.0

# Mengubah label ke format one-hot encoding
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)

# Menampilkan salah satu contoh gambar
plt.imshow(x_train[0].reshape(28, 28), cmap='gray')
plt.title(f"Label: {class_names[np.argmax(y_train[0])]}")
plt.show()
```

membangun model CNN

```python
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')
])

# Kompilasi model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Menampilkan ringkasan model
model.summary()
```

```python
# Melatih model
history = model.fit(x_train, y_train, epochs=10, batch_size=128, validation_data=(x_test, y_test))
```

```python
# Evaluasi model
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=2)
print(f"Akurasi model pada data uji: {test_acc*100:.2f}%")

# Melakukan prediksi pada satu gambar
predictions = model.predict(x_test)
sample_index = 0
predicted_label = np.argmax(predictions[sample_index])
true_label = np.argmax(y_test[sample_index])

print(f"Prediksi model untuk gambar ke-{sample_index}: {class_names[predicted_label]}")
print(f"Label sebenarnya: {class_names[true_label]}")
```
